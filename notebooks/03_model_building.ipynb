```python
listings = data.load("listings")
```

```python
homes_sort = listings.sort_values(by = 'date_sold', ascending = True).reset_index(drop = True)
```

```python
from sklearn.ensemble import HistGradientBoostingRegressor

model = HistGradientBoostingRegressor(random_state = 0)
```

```python
train_prop = 0.9

n = int(len(homes_sort) * train_prop)
```

```python
X_train = features.off_market_features(homes_sort)[:n]
y_train = np.log(homes_sort['sold_price'][:n])
```

```python
model.fit(X_train, y_train)
```

```python
model.score(X_train, y_train)
```

```python
X_valid = features.off_market_features(homes_sort)[n:]
y_valid = homes_sort['sold_price'][n:]
```

```python
preds = model.predict(X_valid)
```
